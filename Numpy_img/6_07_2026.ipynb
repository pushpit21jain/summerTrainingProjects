```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
img = Image.open("pushpit.png")
img = np.array(img)
plt.imshow(img)
plt.axis()
plt.show()
img[: , : , 1]=np.clip(img[ :, :, 1] + 100, 101, 100  )
plt.imshow(img)
plt.axis()
plt.show()
img[:200 , :200 , 0]=np.clip(img[ :200, :200, 0] + 255, 101, 100  )
plt.imshow(img)
plt.axis()
plt.show()
img[:200 , :200 , 2]=np.clip(img[ :200, :200, 1] + 255, 101, 100  )
plt.imshow(img)
plt.axis()
plt.show()
```
