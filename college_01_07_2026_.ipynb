```python
# cash balance help pasword_change
#ATM managmet system

atm_users = {
    "user123": {
        "pin": 1234,
        "balance": 5000,
        "name": "Alex"
    },
    "user567": {
        "pin": 9876,
        "balance": 12500,
        "name": "Sam"
    }
}
def operation():
  userId = input("Enter userId")

  if userId not in atm_users:
    print("Error user not found")
    return

  print("1 - Balance" , "2- withdraw" , "3 - help" , "4 - pin change")
  operation = int (input("Enter operation"))

  userPin = int(input("ENTER PIN"))
  if userPin != atm_users[userId]["pin"]:
      print("invalid pin")
      return

  if operation == 1:
    print(atm_users[userId]["balance"])
    return
  elif operation ==2:
    amount = int(input("Amount to withdraw"))
    if amount > atm_users[userId]["balance"]:
      print("low balance")
    else:
      atm_users[userId]["balance"] = atm_users[userId]["balance"] -amount
      print(atm_users[userId]["balance"])
  elif operation == 3 :
    Querry = input("Enter your Querry")
  elif operation == 4 :
    new_pin = int(input("Enter new pin"))
    atm_users[userId]["pin"] = new_pin
  return

operation()

```
