```python
'''
Q1. Student Information System
Create a Student class with attributes: Name, Roll Number, Marks (3 subjects).
Methods: get_data(), display_data(), calculate_percentage().
'''
class student:
  name =""
  roll_no = 0
  Maths:0
  Physics:0
  Chemistry:0

  def __init__(a,name,roll_no,Maths,Physics,Chemistry):
    a.name = name
    a.roll_no=roll_no
    a.Maths = Maths
    a.Physics=Physics
    a.Chemistry = Chemistry


  def display_data(a):
    print(a.name)
    print(a.roll_no)
    print(a.Maths)
    print(a.Physics)
    print(a.Chemistry)

  def percentage(a):
    print(f"Percentage of {a.name} is {((a.Chemistry+a.Maths+a.Physics)/3)}")

s1=student("Pushpit",123,99,100,97)
s1.percentage()
```

```python
'''Q2. Bank Account
Create a BankAccount class with Account Holder Name, Account Number, Balance.
Methods: deposit(), withdraw(), display_balance().
Conditions: Deposit must be positive and withdrawal cannot exceed balance.
'''

class account():
  account_Holder_Name=""
  account_number = 0
  balance=0
  def __init__(a,account_Holder_Name,account_number,balance):
    a.account_Holder_Name=account_Holder_Name
    a.account_number=account_number
    a.balance=balance

  def deposit(a,amount):
    a.balance =a.balance+amount
    print(f"Balance in {a.account_number} is {a.balance}")
  def withdraw(a,amount):
     if amount < a.balance:
      a.balance =a.balance-amount
      print(f"Balance in {a.account_number} is {a.balance}")
     else:
      print("Low Balance")

  def display_balance(a):
    print(f"Balance in {a.account_number} is {a.balance}")

c1 = account("Pushpit",12334,2000)
c1.withdraw(1000)
```

```python
'''Q3. Employee Salary Calculator
Create an Employee class with Employee ID, Name, Basic Salary.
Methods to calculate HRA (20%), DA (10%), Gross Salary, and display details.
'''

class employee:
  employee_number = 0
  employee_name = ""
  employee_salary = 0
  def __init__(self, employee_name, employee_number, employee_salary):
    self.employee_name = employee_name
    self.employee_number = employee_number
    self.employee_salary = employee_salary

  def hra(self):
    return self.employee_salary * 0.2

  def da(self):
    return self.employee_salary * 0.1

  def gross_sal(self):
    gross_salary = self.employee_salary + self.hra() + self.da()
    return gross_salary

  def display(self):
    print(f"Employee Name: {self.employee_name}")
    print(f"Employee ID: {self.employee_number}")
    print(f"Basic Salary: {self.employee_salary}")
    print(f"HRA (20%): {self.hra()}")
    print(f"DA (10%): {self.da()}")
    print(f"Gross Salary: {self.gross_sal()}")

e1 = employee("pushpit", 123, 10000)

e1.display()
```

```python
'''Q4. Rectangle Operations
Create a Rectangle class with Length and Breadth.
Methods: area(), perimeter(), check_square().
'''

class rec():
  def __init__(a,lenght,breath):
    a.lenght=lenght
    a.breath=breath
  def area(a):
    return a.lenght*a.breath
  def perimeter(a):
    return 2*(a.lenght+a.breath)
  def check_square(a):
    if a.lenght == a.breath:
      return True
    else:
      return False
  def output(a):
    print(rec.area(a))
    print(rec.perimeter(a))
    print(rec.check_square(a))

r1=rec(2,2)
r1.output()
```

```python
'''Q5. Vehicle Management System
Create a Vehicle base class with Brand and Model.
Create Car (Fuel Type) and Bike (Engine Capacity) child classes.
Override display() in child classes.
'''

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display(self):
        return f"Brand: {self.brand}, Model: {self.model}"

class Car(Vehicle):
    def __init__(self, brand, model, fuel_type):
        super().__init__(brand, model)
        self.fuel_type = fuel_type

    def display(self):
        base_display = super().display()
        return f"{base_display}, Fuel Type: {self.fuel_type}"

class Bike(Vehicle):
    def __init__(self, brand, model, engine_capacity):
        super().__init__(brand, model)
        self.engine_capacity = engine_capacity

    def display(self):
        base_display = super().display()
        return f"{base_display}, Engine Capacity: {self.engine_capacity}cc"


vehicle1 = Vehicle("car", "xyz")
print(f"Vehicle: {vehicle1.display()}")
car1 = Car("Toyota", "LC", "Petrol")
print(f"Car 1: {car1.display()}")
bike1 = Bike("Honda", "Activa", 125)
print(f"Bike 1: {bike1.display()}")
```

```python
'''Q6. Person → Teacher → Principal
Use multilevel inheritance.
Person: Name, Age
Teacher: Subject
Principal: School Name
Display all details using Principal object.
'''


class person:
  name=""
  age=0
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def display(self):
    print(self.name)
    print(self.age)

class teacher(person):
  subject=""
  def __init__(self,subject,name,age):
    super().__init__(name,age)
    self.subject=subject

  def display(self):
    print(self.subject,self.name,self.age)



class principal(teacher):
  school_name=""

  def __init__(self,school_name,name,age,subject):
    super().__init__(subject,name,age)
    self.school_name=school_name
  def display(self):
    print(self.school_name,self.subject,self.name,self.age)

p1 = principal("ABC","Pushpit",21,"Maths")
p1.display()
```
